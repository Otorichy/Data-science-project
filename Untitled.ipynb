```python
// The correlation value is -0.064 which indicates a weak negative correlation between the two variables Age and CGPA. 
    This shows that age has no effect on CGPA//
Now lets check to see if its statistically significant using the pvalues
```

```python
from scipy.stats import pearsonr
r,p = pearsonr(df['Age'], df['CGPA'])
print('Correlation:', r)
print('P-value:', p)
```
